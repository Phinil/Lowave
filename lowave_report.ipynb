# 1. Introduktion

Denna brief report sammanfattar arbetet med vår produkt Lowave, som  utvecklades utifrån en design thinking-process. 
Syftet är att beskriva vilka aktiviteter och verktyg som genomfördes i varje steg och hur dessa bidrog till produktens slutresultat. 

Projektet utgick från ett Small Web-perspektiv där vi fokuserade på hur mindre musikaktörer kan få bättre möjligheter att bli upptäckta och bygga relationer med sina lyssnare. 
Genom processen undersöktes hur musikupptäckt kan göras mer personlig, utforskande och mindre beroende av popularitetsdrivna flöden.

Rapporten beskriver även hur dataförberedelse, dataanalys och maskininlärning användes i utvecklingen av produkten, samt hur insikter från varje design-thinking del bidrog till det slutliga resultatet. 


# 2. Design Thinking-processen
Projektet genomfördes iterativt genom flera sprintar, där design thinking användes för att förstå mindre artisters behov, definiera problemet, utveckla idéer, skapa prototyper och förbättra lösningen utifrån feedback. Nya insikter från intervjuer, prototyper och tester påverkade både problemförståelsen och den fortsatta utvecklingen. Fokus låg på små oberoende musikartister och deras svårigheter att bli upptäckta, nå rätt lyssnare och bygga relationer med sin publik på dagens musikplattformar.

## 2.1 Empathize
I empathizefasen fokuserade vi på att förstå problem owners situation och behov. Vi genomförde semistrukturerade  [Intervjuer – Sprint 1](Sprint%201/Interviews) & [Intervjuer – Sprint 2](Sprint%202/intervjuer) med mindre musiker och dokumenterade genom anteckningar, inspelningar och sammanställningar. För att analysera materialet använde vi [empathy maps](Sprint%201/empathy-maps) och [What, How, Why-analys](Sprint%201/What%2C%20how%20why%3F%20(12%20april)%20). Empathy mapsen hjälpte oss att få en tydlig bild av vad artisterna tycker, tänker och behöver. En insikt var att många små artister upplever en konflikt mellan att skapa musik och att behöva anpassa sig till algoritmer, trender och marknadsföring för att synas. En av våra intervjupersoner nämnde detta ”Alla anpassar nog sin musik för lyssningar. Vill ej göra det, men har gjort det lite”. Vissa upplevde också att streams och visningar inte automatiskt leder till en större publik. 
Genom what, how, why blev det tydligt att problemet inte enbart handlar om bristande exponering, utan också om hur dagens musikplattformar är utformade. Plattformarna styr synlighet genom algoritmer och ger begränsade möjligheter till interaktion mellan artist och lyssnare. Den viktigaste insikten från empathizefasen var därför att artisterna inte bara behöver fler visningar, utan mer kontroll över hur de blir upptäckta och hur de bygger relationer med sin publik. Denna insikt tog vårt arbete vidare mot mer rättvis upptäckbarhet och relationer snarare än enbart marknadsföring.

## 2.2 Define
I definefasen sammanställde vi insikterna från intervjuerna och analysverktygen för att avgränsa problemet. Vi använde [POV](Sprint%201/POV) för att formulera problemet ur användarnas perspektiv och [How Might We-frågor](Sprint%201/How%20might%20we%20(12%20april)%20%20). för att omvandla insikterna till designmöjligheter. Vår POV visade att små musikaktörer inte främst saknar en musikplattform eller mer marknadsföring. De behöver istället ett sätt att bli upptäckta och bygga relationer med lyssnare på sina egna villkor. How might we-frågorna hjälpte oss att fokusera på hur mindre artister kan bli upptäckta utan att vara helt beroende av algoritmer. Samt hur lyssnare enklare kan hitta mindre artister som matchar deras musiksmak.
Detta ledde fram, efter några ändringar, till vårt slutgiltiga problem statement:
” Små oberoende musikartister har svårt att bli upptäckta eftersom dagens plattformar prioriterar etablerade artister och popularitet. Detta gör att ny musik sällan når lyssnare, även om den matchar deras smak.”
Den viktigaste insikten från definefasen var att lösningen behövde kombinera två behov: mer rättvis upptäckbarhet för små artister och bättre möjligheter till interaktion mellan artist och lyssnare. Detta gjorde att vi kunde fokusera på lösningar som inte bara ökar synlighet, utan hjälper artister att nå rätt publik.

## 2.3 Ideate
I ideatefasen tog vi fram möjliga lösningar utifrån de behov och problem som identifierats tidigare. Vi började med [Brainstorming](Sprint%202/Brainstorming(13%20april)), där vi  hjälptes åt att ta fram mängder av idéer för lösningen. Brainstormingen resulterade i 18 lösningsidéer. Därefter använde vi [ABC Idea Sorting](Sprint%202/ABCideasorting(13%20april).jpg) för att sortera idéerna utifrån relevans, genomförbarhet och hur väl de svarade mot problem owners behov. 
Genom A/B/C sorteringen blev det tydligt att vissa idéer var intressanta men för omfattande eller mindre kopplade till kärnproblemet. Vi valde därför att kombinera de idéer som bäst hjälpte våra problem owners: att hjälpa lyssnare hitta mindre artister på ett enkelt och mer personligt sätt. 
Den valda riktningen blev en musikplattform där lyssnare kan upptäcka independent artister genom mood, genre och prompts, samt få rekommendationer som matchar deras musiksmak. Small web-perspektivet påverkade också idéutvecklingen, eftersom vi ville skapa en mer personlig, mer communityliknande och mindre algoritmstyrd plattform. Den viktigaste insikten från ideatefasen var att lösningen inte bara skulle hjälpa artister att synas mer, utan hjälpa dem att nå rätt lyssnare på ett sätt som känns mer autentiskt och relevant. Detta gjorde att vi gick vidare med en prototyp som fokuserade på discovery, musikpreviews och enkel interaktion.

## 2.4 Prototype
I prototypefasen gjorde vi lösningen visuell genom flera typer av prototyper. Vi skapade tidiga versioner på papper, i Figma och genom kod för att snabbt kunna testa både funktioner, design och användarupplevelse: [Prototyp 1](Sprint%202/Prototyp%201). Prototypandet skedde iterativt, vilket innebar att feedback från PO och gruppens egna tester påverkade vilka funktioner som utvecklades vidare eller byttes ut eller eventuellt togs bort. Vi använde även i denna fasen brainstorming för nya eventuella funktioner i våra prototyper. 
Prototypen utvecklades till en musikplattform där användare kan upptäcka independent artister genom genre och moodbaserade rekommendationer. Under arbetet implementerades flera funktioner, bland annat musikpreviews, feed, spar- och kommenteringsfunktioner, sökfunktion och användning av ISRC-kod.
Prototypandet gjorde det tydligt vilka delar av idén som var mest relevanta och värdefulla för användarna. Musikpreviews framstod som väldigt viktiga, eftersom de gjorde det enkelt att direkt bedöma om en låt var intressant. Samtidigt blev det tydligt att community- och feedfunktionerna behövde förtydligas. Den viktigaste insikten från prototypfasen var att lösningen behövde vara enkel men proffsig och lätt att förstå. Eftersom vi vill att användaren snabbt ska kunna hitta ny musik, medan artisten får en ärlig chans till synlighet utan att bli för låst av algoritmer.

## 2.5 Test
I testfasen utvärderade vi prototypen genom användartester, [testintervjuer](Sprint%203/Interviews) och [interna tester](Sprint%204/testing%20recommendation) av rekommendationsfunktionen. Syftet var att undersöka om plattformen var tydlig, relevant och användbar, samt om rekommendationerna gav rimliga resultat utifrån användarens val av exempelvis mood och genre.
Användartesterna visade att deltagarna var positiva till grundidén och uppskattade fokuset på mindre artister. Previewfunktionen lyftes fram som väldigt bra, eftersom den gjorde det lättare att snabbt upptäcka ny musik. Vibesökningen upplevdes också som en stark funktion, eftersom den lät användaren söka musik utifrån känsla snarare än enbart artistnamn eller genre. Testerna visade samtidigt flera förbättringsområden. Communityflödet upplevdes som oklart och behövde mer visuellt innehåll och tydligare interaktionsmöjligheter, exempelvis kommentarer, trådar eller möjligheten att pinna inlägg. Discoveryfunktionen behövde fler genrer och kategorier och ISRC-koden behövde förklaras tydligare för artister som inte känner till den. Vid interna tester märkte vi även att fri textsökning ibland gav för breda eller irrelevanta rekommendationer. Därför ändrades funktionen mot mer styrda val, där användaren väljer mood, genre och andra låtegenskaper.
Den viktigaste insikten från testfasen var att plattformen fungerade bäst när den gav användaren tydliga val och interaktionsmöjligheter samt tillgång till musikpreviews. Testerna ledde därför till förbättringar av rekommendationsfunktionen, communityflödet och prototypens tydlighet. Sammantaget gjorde testfasen att lösningen blev mer användarcentrerad och bättre anpassad till vårt mål. 



# 3. Dataförberedelse och analys
Vi använde ett dataset från Kaggle med cirka 114 000 låtar innehållande ljudegenskaper som tempo, energi och akustiskhet. En [explorativ dataanalys (EDA)](Sprint%202/EDA) genomfördes för att förbättra datakvaliteten och identifiera mönster.
Vi rensade bort irrelevanta kolumner och standardiserade kategoriska variabler. Inga exakta dubbletter hittades, men flera rader med samma track_id förekom, vilket behölls då dessa inte var identiska datapunkter. Tre rader med saknade centrala värden (artist, album eller låtnamn) togs bort.
Analysen visade att de flesta låtar har låg till medelhög popularitet, medan få når mycket hög popularitet. Vissa genrer var generellt mer populära än andra. En korrelationsanalys visade även samband mellan vissa variabler, exempelvis mellan energi och ljudstyrka.

# 4. Maskininlärning
Vi använde “unsupervised” maskininlärning för att gruppera låtar baserat på deras ljudegenskaper. Följande features valdes: danceability, energy, liveness, tempo, instrumentalness, acousticness och valence. Dessa standardiserades innan modellträning.
Tre metoder testades: KMeans, DBSCAN och agglomerativ klustring. Samtliga gav liknande silhouette score, vilket indikerar att datan saknar tydliga kluster, något som är rimligt för musikdata. 
[Workshop: Unsupervised Machine Learning](Sprint%203/Ml%20%26%20unsupervisedlearning/unsupervised_ml_workshop.ipynb)

Vi valde KMeans eftersom metoden ger tydliga och kontrollerbara kluster som är enkla att använda i en rekommendationsmodell. Antalet kluster bestämdes med elbow method och sattes till fem, vilka representerar olika musikaliska “vibes”. För att undersöka modellens begränsningar testades den även på ett syntetiskt dataset (two moons), där KMeans presterade sämre på grund av dess antagande om sfäriska kluster. Detta visar att val av metod bör anpassas efter datans struktur.
I vår prototyp har vi med hjälp av KMeans byggt ett system där användare kan välja en genre och även via “sliders” önska ungefär vilka features låtarna ska ha. Vår KMeans modell väljer sedan vilket kluster som passar bäst efter användarens önskemål. Därefter avgör vår [kod (prototyp 15)](Sprint%204/kod_prototyp15) vilken låt i klustret som är närmast användarens input (seed-låt) och tar också fram låtar som liknar seed-låten.
Eftersom musik är subjektivt finns det ingen perfekt modell, vilket också märktes i tidiga versioner av applikationen där vissa rekommendationer upplevdes som oväntade. Denna rekommendationsfunktion med hjälp av KMeans och seed-låtar tror vi bäst passar våra problem-owners behov utifrån de olika alternativ vi testat. 



# 6. Slutresultat

Det slutliga resultatet blev Lowave,  som är en interaktiv musikplattform inspirerad av Small Web-perspektivet. Plattformen utvecklades med fokus på discovery, community och mer personlig musikupptäckt. Målet var att underlätta för vår problem owner, det vill säga mindre musikaktörer, att nå ut med sin musik, samtidigt som lyssnaren får möljlighet att upptäcka nya artister och musik på ett mer utforskande och personligt sätt. 

En central del av produkten blev ett discovery-flöde där användare kan utforska nya låtar från mindre artister genom moods, genrer och randomiserade rekommendationer istället för popularitetsbaserade flöden. Tanken bakom detta var att skapa bättre möjligheter för mindre musikaktörer att bli upptäckta utan att först behöva konkurrera genom stora algoritmiska system.

Produkten utvecklades även med flera community-funktioner där användare kan dela musik, skapa inlägg och diskutera både låtar och musikupptäckter tillsammans med andra användare. Plattformen innehåller även ett koncept för community-rum där användare och artister kan interagera kring specifika moods, genrer och musikintressen i egna “rum”. Fokuset med detta låg på att skapa en mer personlig relation mellan artist och lyssnare istället för att enbart fokusera på streams och marknadföring. 

För själva musikaktörerna utvecklades även ett enkelt uppladdningsflöde baserat på ISRC-koder, vilket minskade både tekniska och administrativa hinder för att bli sökbar i plattformen. Artisterna kan även få direkt feedback från lyssnare kopplat till sina låtar vilket stödjer deras efterfrågan på mer direkt interaktion. 

En annan viktig del av slutresultatet blev rekommendationssystemet där användare kan “mixa fram en känsla” genom sliders kopplade till olika musikaliska egenskaper och moods. Funktionen kombinerades med clustering för att skapa mer sammanhängande rekommendationer baserade på musikaliska mönster istället för enbart popularitet eller användardata. Detta gör musik upptäckten enklare och mer transparent.  

Det slutliga resultatet förändrades kontinuerligt genom prototyper, användartester och feedback under design thinking-processen, där flera funktioner utvecklades och vidareutvecklades baserat på både användarinsikter och tekniska begränsningar. 

# 7. Reflektion

Projektet visade hur stor påverkan design thinking-processen hade på utvecklingen av vår produkt. Genom intervjuer, design tools och användarinsikter förändrades flera delar av vår ide under arbetets gång. Tidigt låg fokus främst på musikupptäckt och rekommendationer, men genom research blev det tydligt att relationen mellan artist och lyssnare var minst lika viktig som själva discovery-delen. Detta ledde till att community-funktioner fick en större roll i det slutliga resultatet.

En av de främsta lärdomarna från projektet var hur svårt det är att skapa alternativa rekommendationssystem inom musik utan att falla tillbaka på samma logik som traditionella plattformar. Vi upptäckte snabbt att det finns en tydlig utmaning mellan Small Web-idén och beroendet av stora plattformar, API:er och musikdata. Samtidigt visade projektet och vår slutgiltiga produkt att det fortfarande går att skapa mer utforskande och mindre popularitetsstyrda rekommendationer, genom att bland annat fokusera på musikens egenskaper,  istället för användardata och klickhistorik.

Arbetet med maskininlärning och clustering gav också en större förståelse för hur man kan skapa en modell trots musikens subjektivitet. Det blev tydligt att det inte finns någon perfekt modell för rekommendationer, eftersom musiksmak varierar mellan olika personer och genrer ofta överlappar varandra. Genom att testa olika modeller, som visade både styrkor och begräsningar, kunde vi välja modell utifrån produktens faktiska syfte och användarupplevelse. 

En annan viktig insikt var hur mycket prototyper och användartester påverkade produktens utveckling, och hur viktigt det var för att uppnå ett slutresultat som var välarbetat och genomtänkt. Processen blev på så sätt betydligt mer iterativ än vad vi initialt förväntade oss.

Sammanfattningsvis bidrog design thinking-processen till att projektet utvecklades stegvis genom research, prototyper, tester och reflektion. Istället för att utveckla en färdig lösning direkt hjälpte processen oss att kontinuerligt omvärdera både problem, funktioner och tekniska val utifrån användarnas behov och projektets vision.
